# Polynomial Linear Regression with scikit-learn

We fit a **polynomial** relationship between one input $x$ and the target $y$ by first creating polynomial features ($x$, $x^2$, $x^3$, ...) and then using **ordinary least squares** on those features. The model is still **linear in the coefficients**, so it is a form of linear regression. We use **PolynomialFeatures** and **LinearRegression** from scikit-learn.

## Imports

**numpy** for arrays and random data, **matplotlib** for plotting, **sklearn.linear_model.LinearRegression** for the model, **sklearn.preprocessing.PolynomialFeatures** to build polynomial terms, and **sklearn.metrics.mean_squared_error** for evaluation.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.metrics import mean_squared_error
```

## Generate synthetic data

We create data from a **quadratic** relationship $y = 2 + 0.5 x - 0.3 x^2 + \varepsilon$ with Gaussian noise $\varepsilon$. This is nonlinear in $x$, but linear in the coefficients once we use features $[1, x, x^2]$.

```python
np.random.seed(42)
n_samples = 80
x = np.linspace(-2, 3, n_samples)
# True model: y = 2 + 0.5*x - 0.3*x^2 + noise
y = 2 + 0.5 * x - 0.3 * x**2 + np.random.randn(n_samples) * 0.5
x = x.reshape(-1, 1)  # sklearn expects shape (n_samples, n_features)
y = y.reshape(-1, 1)
```

## Build polynomial features and fit

**PolynomialFeatures(degree=2)** creates a design matrix with columns $[1, x, x^2]$ (including the intercept term by default). We **fit** it on the training input **x** and **transform** to get **X_poly**. Then we fit **LinearRegression** on **X_poly** and **y**. The fitted model is $\hat{y} = \theta_0 + \theta_1 x + \theta_2 x^2$.

```python
degree = 2
poly = PolynomialFeatures(degree=degree, include_bias=True)
X_poly = poly.fit_transform(x)

model = LinearRegression()
model.fit(X_poly, y)
y_pred = model.predict(X_poly)
```

## Inspect coefficients and RMSE

**model.intercept_** is $\theta_0$ and **model.coef_** are $\theta_1, \theta_2, ...$. We compare them to the true coefficients (2, 0.5, -0.3) and report RMSE.

```python
print("Intercept (theta_0):", model.intercept_[0])
print("Coefficients (theta_1, theta_2, ...):", model.coef_[0])
rmse = np.sqrt(mean_squared_error(y, y_pred))
print("RMSE on training data:", round(rmse, 4))
```

## Plot data and fitted curve

We plot the raw points and the predicted curve. For a smooth curve we evaluate the model on a dense grid of **x_plot** after transforming it with the same **PolynomialFeatures**.

```python
x_plot = np.linspace(x.min(), x.max(), 200).reshape(-1, 1)
X_plot_poly = poly.transform(x_plot)
y_plot = model.predict(X_plot_poly)

plt.figure(figsize=(8, 5))
plt.scatter(x, y, alpha=0.7, label="Data")
plt.plot(x_plot, y_plot, color="red", linewidth=2, label=f"Fitted polynomial (degree={degree})")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.title("Polynomial Linear Regression (sklearn)")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## Optional: try degree 3

Using **degree=3** adds an $x^3$ term. For our quadratic true model, degree 2 is enough; degree 3 may overfit slightly but can still fit the data well. Try changing **degree** above or re-running with `degree = 3` to compare.

```python
degree3 = 3
poly3 = PolynomialFeatures(degree=degree3, include_bias=True)
X_poly3 = poly3.fit_transform(x)
model3 = LinearRegression()
model3.fit(X_poly3, y)
y_pred3 = model3.predict(X_poly3)
rmse3 = np.sqrt(mean_squared_error(y, y_pred3))
print("Degree 3 - Coefficients:", model3.coef_[0])
print("Degree 3 - RMSE:", round(rmse3, 4))

x_plot = np.linspace(x.min(), x.max(), 200).reshape(-1, 1)
plt.figure(figsize=(8, 5))
plt.scatter(x, y, alpha=0.7, label="Data")
plt.plot(x_plot, model.predict(poly.transform(x_plot)), color="red", linewidth=2, label="Degree 2")
plt.plot(x_plot, model3.predict(poly3.transform(x_plot)), color="green", linewidth=2, label="Degree 3")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.title("Polynomial regression: degree 2 vs 3")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
