````python
import numpy as np

class Perceptron:
    def __init__(self, lr=0.1, epochs=10):
        self.lr = lr
        self.epochs = epochs

    def fit(self, X, y):
        self.weights = np.zeros(X.shape[1])
        self.bias = 0

        for _ in range(self.epochs):
            for i in range(len(X)):
                linear_output = np.dot(X[i], self.weights) + self.bias
                y_pred = self.activation(linear_output)

                # Update rule
                update = self.lr * (y[i] - y_pred)
                self.weights += update * X[i]
                self.bias += update

    def activation(self, x):
        return 1 if x >= 0 else 0

    def predict(self, X):
        return [self.activation(np.dot(x, self.weights) + self.bias) for x in X]
````

### Perceptron `fit` lifecycle

The method matches this loop: **initialize** → **repeat for each epoch** → **for each training sample** compute the linear score, threshold to a class, then **nudge** `w` and `b` only when `y` and the prediction differ.

| Step | In code | Meaning |
|------|---------|--------|
| 1 | `weights = 0`, `bias = 0` | Start from the origin in weight space |
| 2 | outer `for _ in range(epochs)` | Pass over the dataset multiple times |
| 3 | inner `for i in range(len(X))` | One sample `X[i]` with label `y[i]` |
| 4 | `linear_output = dot(X[i], w) + b` | Weighted sum plus bias |
| 5 | `y_pred = activation(linear_output)` | `1` if score ≥ 0, else `0` |
| 6 | `update = lr * (y[i] - y_pred)` | Zero if correct; ±`lr` if wrong |
| 7 | `weights += update * X[i]`, `bias += update` | Move the hyperplane toward fixing the mistake |

**ASCII sketch** (same control flow as your `for` loops):

```
                    ┌──────────────────────────────────┐
                    │  fit: w ← 0,  b ← 0              │
                    └─────────────────┬────────────────┘
                                      ▼
                    ┌──────────────────────────────────┐
              ┌────►│  for each epoch                  │
              │     └─────────────────┬────────────────┘
              │                       ▼
              │     ┌──────────────────────────────────┐
              │     │  for each sample i               │
              │     └─────────────────┬────────────────┘
              │                       ▼
              │           z = w·X[i] + b
              │                       ▼
              │           ŷ = 1 if z ≥ 0 else 0
              │                       ▼
              │           u = lr * (y[i] − ŷ)
              │                       ▼
              │           w ← w + u·X[i],  b ← b + u
              │                       │
              └───────────────────────┘   (next i, then next epoch)
```

**Mermaid** (renders in VS Code / GitHub / some Jupyter setups):

```mermaid
flowchart TB
    A([fit starts]) --> B["w ← 0, b ← 0"]
    B --> C{more pairs epoch / sample left?}
    C -->|yes| D["z = w·X[i] + b"]
    D --> E["ŷ = 1 if z ≥ 0 else 0"]
    E --> F["u = lr · (y[i] − ŷ)"]
    F --> G["w += u·X[i], b += u"]
    G --> C
    C -->|no| Z([fit ends])
```

When the prediction is already correct, `y[i] - y_pred` is `0`, so **no update** happens for that sample.

````python
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

y = np.array([0, 0, 0, 1])
````

````python
model = Perceptron()
model.fit(X, y)

print(model.predict(X))
````

````python
y_xor = np.array([0, 1, 1, 0])
model.fit(X, y_xor)
print(model.predict(X))
````

````python
from sklearn.neural_network import MLPClassifier

model = MLPClassifier(hidden_layer_sizes=(2,), max_iter=1000)
model.fit(X, y_xor)

print(model.predict(X))
````
