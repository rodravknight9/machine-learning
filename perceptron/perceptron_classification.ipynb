Problem to solve: “Given hours studied → pass or fail?”

```python
import numpy as np

X = np.array([[1], [2], [3], [4], [5], [6]])
y = np.array([0, 0, 0, 1, 1, 1])  
```

```python
from sklearn.linear_model import Perceptron

model = Perceptron()
model.fit(X, y)

print(model.predict([[2.5]])) 
print(model.predict([[5.5]]))  
```
